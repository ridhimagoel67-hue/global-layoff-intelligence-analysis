# Exploratory Data Analysis (EDA)

## Question 1:
How have global layoff trends evolved from 2020–2026?

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    "../data/processed/layoffs_cleaned.csv",
    parse_dates=["date", "date_added"]
)

df.shape
```

```python
layoff_events_by_year = df.groupby("year").size()

layoff_events_by_year
```

```python
df[df["year"] == 2021].head()
```

```python
df[df["year"] == 2021].shape
```

```python
layoffs_by_year = df.groupby("year")["total_laid_off"].sum()

layoffs_by_year
```

```python
avg_layoff_size = df.groupby("year")["total_laid_off"].mean()

avg_layoff_size
```

```python
quarterly_events = df.groupby(["year", "quarter"]).size()

quarterly_events
```

```python
year_summary = pd.DataFrame({
    "Layoff_Events": df.groupby("year").size(),
    "Employees_Laid_Off": df.groupby("year")["total_laid_off"].sum(),
    "Average_Layoff_Size": df.groupby("year")["total_laid_off"].mean()
})

year_summary
```

# Key Insights

## 1. Layoff activity peaked in 2023.

2023 recorded the highest number of layoff events (1,387) and the highest number of employees laid off (264,320).
This indicates that 2023 was the most severe year for workforce reductions in the dataset.

## 2. Two major layoff waves are visible.

The first wave occurred in 2020, driven by the COVID-19 pandemic and economic uncertainty.
The second and larger wave occurred during 2022–2023, likely reflecting the post-pandemic market correction, funding slowdown, and restructuring across industries.

## 3. Layoff activity declined after 2023.

The number of layoff events fell from 1,387 in 2023 to 633 in 2024.
This suggests that workforce reduction activity began stabilizing after the peak period.

## 4. Average layoff size increased over time.

Average layoff size increased from approximately 171 employees per event in 2020 to over 312 employees per event in 2023.
Although later years show even higher averages, these years may be incomplete and should be interpreted cautiously.

## 5. The largest concentration of layoff events occurred in Q1 2023.

Q1 2023 alone recorded 586 layoff events, the highest quarterly value in the dataset.
This suggests that many organizations initiated workforce reductions at the beginning of 2023.

Which industries experienced the highest layoff activity and workforce reductions?

# Key Insights

## 1. Finance recorded the highest number of layoff events.

The Finance industry reported 527 layoff events, the highest among all industries.
This indicates that workforce restructuring was widespread across financial organizations during the study period.

## 2. Retail and Hardware experienced the largest workforce reductions.

Retail companies laid off approximately 106,706 employees.
Hardware companies laid off approximately 106,157 employees.
These industries experienced the greatest overall workforce impact despite having fewer layoff events than Finance.

## 3. Hardware had the most severe layoffs.

Hardware recorded an average layoff size of approximately 1,930 employees per event, the highest among all industries.
This suggests that layoffs in the Hardware sector were less frequent but significantly larger when they occurred.

## 4. Consumer and Retail industries were consistently affected across multiple metrics.

Both industries ranked highly in:
Number of layoff events
Total employees laid off
Average layoff size
This indicates sustained workforce pressure across these sectors.

## 5. Industry impact differs depending on the metric used.

Finance led in layoff frequency.
Retail and Hardware led in total workforce reductions.
Hardware led in layoff severity (average employees affected per event).
Therefore, no single metric fully captures industry-level layoff risk.

Which company stages are most vulnerable to layoffs?

# Key Insights

## 1. Post-IPO companies experienced the highest layoff activity and workforce reductions.

Post-IPO firms recorded 1,057 layoff events, the highest among all company stages.
These companies accounted for approximately 582,668 employees laid off.
They also recorded the highest average layoff size (777 employees per event).
This indicates that publicly listed companies were the primary drivers of workforce reductions during the study period.

## 2. Growth-stage startups experienced frequent layoffs but at smaller scales.

Series B (484 events), Series C (445 events), and Series D (358 events) reported high layoff activity.
However, their average layoff sizes ranged between 97 and 108 employees per event, substantially lower than Post-IPO firms.
This suggests that growth-stage startups frequently adjusted workforce levels but typically conducted smaller layoffs.

## 3. Acquired companies underwent substantial restructuring.

Acquired firms recorded 398 layoff events and approximately 73,178 employees laid off.
Their average layoff size was approximately 279 employees per event, indicating that mergers and acquisitions were often accompanied by workforce consolidation.

## 4. Early-stage startups were the least affected.

Seed-stage companies recorded 147 layoff events and only 2,751 employees laid off.
Average layoff size was approximately 57 employees per event, the lowest among major stages.
This suggests that early-stage startups contributed relatively little to overall workforce reductions.

## 5. Layoff severity generally increased with company maturity.

Mature stages such as Post-IPO, Acquired, Private Equity, and Subsidiary companies reported significantly larger layoffs per event.
In contrast, Series A, Series B, Series C, and Seed companies reported much smaller workforce reductions on average.
This indicates that larger and more mature organizations tend to implement layoffs on a much greater scale when restructuring occurs.# 

## Question 5
Is there a relationship between funds raised and the magnitude of layoffs?

```python
df["funds_raised"].describe()
```

```python
df[["funds_raised", "total_laid_off"]].corr()
```

```python
df[
    ["company", "funds_raised", "total_laid_off", "stage"]
].sort_values(
    by="funds_raised",
    ascending=False
).head(15)
```

```python
df.groupby("stage")["funds_raised"] \
  .mean() \
  .sort_values(ascending=False)
```

```python
df["funding_group"] = pd.cut(
    df["funds_raised"],
    bins=[0, 50, 200, 500, 1000, 2000, 200000],
    labels=[
        "0-50M",
        "50-200M",
        "200-500M",
        "500M-1B",
        "1B-2B",
        "2B+"
    ]
)
```

```python
funding_summary = df.groupby("funding_group").agg({
    "total_laid_off": "sum",
    "company": "count"
})

funding_summary
```

# Key Insights

## 1. Funding raised showed only a weak relationship with layoffs.

The correlation between funds raised and employees laid off was approximately 0.12.
This indicates a very weak positive relationship.
Companies that raised more capital were not necessarily protected from workforce reductions.

## 2. Both highly funded and lightly funded companies conducted significant layoffs.

Companies that raised less than $50 million accounted for approximately 168,731 employees laid off.
Companies that raised more than $2 billion accounted for approximately 166,199 employees laid off.

This suggests that layoffs affected organizations across the entire funding spectrum.

## 3. The highest number of layoff events occurred among moderately funded companies.

Funding Group	Layoff Events
50–200M	           1,147
0–50M	            934
200–500M	        891
Mid-sized companies represented the largest share of layoff announcements.
These firms may have faced greater pressure to balance growth expectations with profitability goals.

## 4. Large funding rounds did not guarantee workforce stability.

Several highly funded companies, including major Post-IPO firms, still conducted large-scale layoffs.
This indicates that workforce reductions were often driven by strategic restructuring, market conditions, or profitability objectives rather than funding availability alone.

## 5. Layoffs were distributed across all funding categories.

Significant workforce reductions were observed in every funding bracket.
No funding category appeared immune to layoffs, suggesting that broader economic and business factors played a larger role than capital availability.

## Question 6
Which companies conducted the largest layoffs, and what characteristics do they share?

```python
company_layoffs = (
    df.groupby("company")["total_laid_off"]
      .sum()
      .sort_values(ascending=False)
)

company_layoffs.head(20)
```

```python
company_summary = (
    df.groupby("company")
      .agg({
          "total_laid_off":"sum",
          "funds_raised":"first",
          "stage":"first",
          "industry":"first",
          "country":"first"
      })
      .sort_values(
          by="total_laid_off",
          ascending=False
      )
)

company_summary.head(20)
```

# Key Insights

## 1. Amazon recorded the highest number of employees laid off among all companies in the dataset, followed by Intel, Meta, Oracle, and Microsoft.

Amazon reported approximately 58,124 layoffs.
Intel reported approximately 43,115 layoffs.
Meta reported approximately 35,700 layoffs.
Oracle and Microsoft each reported more than 30,000 layoffs.

## 2. Most of the companies with the largest layoffs were Post-IPO organizations.

The majority of the top-ranked companies were publicly listed firms.
This indicates that mature organizations were more likely to implement large-scale workforce restructuring than early-stage startups.

## 3. The United States dominated the list of companies with the largest layoffs.

Most of the top companies, including Amazon, Meta, Microsoft, Oracle, Dell, Cisco, Google, Tesla, PayPal, Uber, and HP, were based in the United States.
This aligns with earlier findings that the United States experienced the highest overall layoff activity.

## 4. Technology-oriented industries were heavily represented among the largest layoffs.

Hardware companies such as Intel, Dell, HP, and Micron appeared frequently among the top layoff organizations.
Consumer and digital technology firms such as Meta and Google also reported substantial workforce reductions.
This suggests that the technology ecosystem experienced significant restructuring during the study period.

## 5. High funding levels did not prevent workforce reductions.

Several companies with substantial funding histories, including Meta, Tesla, Uber, and Amazon, still conducted major layoffs.
This reinforces the earlier finding that funding availability alone was not a strong predictor of workforce stability.

## 6. Large workforce reductions were concentrated among mature, global organizations.

Companies with the largest layoffs generally operated at a global scale, employed large workforces, and had established market positions.
Layoffs were therefore concentrated among mature enterprises rather than smaller startups.
