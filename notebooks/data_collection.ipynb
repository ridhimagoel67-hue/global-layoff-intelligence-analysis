```python
import re

match = re.search(r'urlWithParams:\s*"([^"]+)"', response.text)

encoded_url = match.group(1)

decoded_url = encoded_url.encode().decode("unicode_escape")

full_url = "https://airtable.com" + decoded_url

print(full_url[:1000])
```

```python
test_response = requests.get(full_url)

print("Status:", test_response.status_code)
print("Content-Type:", test_response.headers.get("content-type"))
print("Length:", len(test_response.content))
```

```python
import requests

session = requests.Session()

embed_url = "https://airtable.com/embed/app1PaujS9zxVGUZ4/shroKsHx3SdYYOzeh?backgroundColor=green&viewControls=on"

page = session.get(embed_url)

print("Embed Status:", page.status_code)

print("Cookies acquired:")
print(session.cookies.get_dict())
```
